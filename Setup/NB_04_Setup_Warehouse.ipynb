```sql
-- DimStation
IF OBJECT_ID('dbo.DimDate', 'U') IS NULL
BEGIN
    CREATE TABLE dbo.DimDate
    (
        DateKey INT,
        FullDate DATE,
        CalendarYear INT,
        CalendarMonth INT,
        MonthName VARCHAR(20),
        QuarterNumber INT,
        DayOfMonth INT,
        DayOfWeekName VARCHAR(20),
        RailPeriod VARCHAR(10)
    );
END;
GO

print 'Checked/Created DimDate'

-- DimRoute
IF OBJECT_ID('dbo.DimRoute', 'U') IS NULL
BEGIN
    CREATE TABLE dbo.DimRoute
    (
        RouteKey BIGINT IDENTITY,
        OriginCRS VARCHAR(3) NOT NULL,
        OriginStationName VARCHAR(200) NOT NULL,
        DestinationCRS VARCHAR(3) NOT NULL,
        DestinationStationName VARCHAR(200) NOT NULL
    );
END;
GO

print 'Checked/Created DimRoute'

-- FactPassengerLoad
IF OBJECT_ID('dbo.FactPassengerLoad', 'U') IS NULL
BEGIN
    CREATE TABLE dbo.FactPassengerLoad
    (
        PassengerLoadKey BIGINT IDENTITY,

        RouteKey BIGINT NOT NULL,
        DateKey INT NOT NULL,

        RailPeriod VARCHAR(10) NOT NULL,
        ServiceDate DATE NOT NULL,

        UID VARCHAR(50) NOT NULL,
        TrainID VARCHAR(50) NOT NULL,

        TotalLoad INT,
        TotalLoadPct DECIMAL(10,2),

        NumberOfCounts INT,
        NumberOfCars INT,

        SilverLoadTimestamp DATETIME2(6),
        GoldLoadTimestamp DATETIME2(6)
    );
END;
GO

print 'Checked/Created FactPassengerLoad'
```
