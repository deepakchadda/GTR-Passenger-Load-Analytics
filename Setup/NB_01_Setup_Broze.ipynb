```python
print("START: Setting up Bronze")
```

```python
spark.sql("""
DROP TABLE IF EXISTS file_watermark
""")
```

```python
spark.sql("""
DROP TABLE IF EXISTS file_audit
""")
```

```python
spark.sql("""
CREATE TABLE IF NOT EXISTS `GTR Passenger Loading - PROD`.LH_Bronze.dbo.file_watermark
(
    rail_period STRING,    
    source_file_name STRING,
    source_last_modified_timestamp TIMESTAMP,
    bronze_table_name STRING,
    last_loaded_timestamp TIMESTAMP,
    load_status STRING
)
USING DELTA
""")

print("✅ file_watermark checked/created")
```

```python
spark.sql("""
CREATE TABLE IF NOT EXISTS `GTR Passenger Loading - PROD`.LH_Bronze.dbo.file_audit
(
    audit_id STRING,
    file_name STRING,
    rail_period STRING,
    bronze_table_name STRING,
    load_start_timestamp TIMESTAMP,
    load_end_timestamp TIMESTAMP,
    row_count BIGINT,
    load_status STRING,
    error_message STRING
)
USING DELTA
""")

print("✅ file_audit checked/created")
```
